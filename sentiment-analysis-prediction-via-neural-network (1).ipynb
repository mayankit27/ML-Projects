```python
#Step-1:- Import Necessary librariea
import numpy as np
import pandas as pd
import re 
import nltk
from nltk.corpus import stopwords
import swifter
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN,Dense,Dropout
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report

#Step-2:- Load and Explore the dataset
data=pd.read_csv("/kaggle/input/imdb-dataset-of-50k-movie-reviews/IMDB Dataset.csv")
print(data.info())
print(data.head())
print(data.isnull().sum())

#Step-3:- Clean and Preprocess the data
stop_words=set(stopwords.words("english"))
def clean_text(text):
    text=re.sub("<[^>]*>","",text)
    text=re.sub("[^a-zA-Z]"," ",text)
    text=text.lower().split()
    text=[word for word in text if word not in stop_words]
    return " ".join(text)
data["cleaned"]=data["review"].swifter.apply(clean_text)
data["label"]=data["sentiment"].map({"positive":1,"negative":0})

#Step-4:- Tokenize the data 
max_words=10000
max_len=150
tokenizer=Tokenizer(num_words=max_words,oov_token="<OOV>")
tokenizer.fit_on_texts(data["cleaned"])
sequences=tokenizer.texts_to_sequences(data["cleaned"])

#Step-5:- Select input and output feature
x=pad_sequences(sequences,maxlen=max_len,padding="post",truncating="post")
y=np.array(data["label"])

#Step-6:- Split the data into training and testing sets
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,shuffle=True)

#Step-7:- Build a Simple Recurrent Neural Network
model=Sequential([
    Embedding(input_dim=max_words,output_dim=256,input_length=max_len),
    SimpleRNN(256,activation="tanh",return_sequences=False),
    Dropout(0.3),
    Dense(128,activation="relu"),
    Dropout(0.3),
    Dense(1,activation="sigmoid")
])

#Step-8:- Compile the model
model.compile(optimizer=Adam(learning_rate=0.001),loss="binary_crossentropy",metrics=["accuracy"])
print(model.summary())

#Step-9:- Train the model
early_stop=EarlyStopping(monitor="val_accuracy",patience=5,restore_best_weights=True,verbose=1)
model.fit(x_train,y_train,epochs=15,batch_size=128,validation_split=0.1,verbose=1)

#Step-10:- Predict the model
y_pred=model.predict(x_test,batch_size=128)
y_pred=(y_pred>0.5).astype("int32")

#Step-11:- Model Evaluation
acc=accuracy_score(y_test,y_pred)
print("Acc: ",acc)
cm=confusion_matrix(y_test,y_pred)
print("CM: ",cm)
cr=classification_report(y_test,y_pred)
print("CR: ",cr)

#Step-12:- Predict sentiment for new data 
sample_review=["It is too good","It is too bad"]
review_sequences=tokenizer.texts_to_sequences(sample_review)
review_pad_sequences=pad_sequences(review_sequences,maxlen=max_len,padding="post",truncating="post")
predictions=model.predict(review_pad_sequences)
predictions=(predictions>0.5).astype("int32")
print("Predictions: ",predictions)
```
